# Natural Scene Classification

Custom CNN versus frozen ImageNet ResNet50V2. Adapted from the collaborative CNN_Image_Classification repository; see ATTRIBUTION.md. Historical validation results are in RESULTS.md. This cleaned notebook has no newly measured results.


```python
import os
from pathlib import Path
from scene import train, evaluate

DATA_DIR = os.environ.get("SCENE_DATA_DIR", "")
RUN_DIR = Path("runs/comparison")
SEED, EPOCHS, IMAGE_SIZE, BATCH_SIZE = 123, 15, 64, 32
RUN_TRAINING = False  # Set True only when you intend to train both networks.
RUN_EVALUATION = False  # Requires saved models and their split manifest.
```

## Training

Point SCENE_DATA_DIR to seg_train, containing the six class folders. Stable content-deduplicated file splits are written before training; both models share them. Raw RGB pixels enter each model. The custom CNN scales to [0,1]; ResNet50V2 scales to [-1,1]. Augmentation runs only during training. No batch-level take/skip splitting or cached augmentation is used.


```python
if RUN_TRAINING:
    if not DATA_DIR:
        raise ValueError("Set SCENE_DATA_DIR first")
    train(DATA_DIR, RUN_DIR, SEED, EPOCHS, IMAGE_SIZE, BATCH_SIZE)
else:
    print("Training disabled; no models trained or weights downloaded.")
```

## Held-out evaluation

Run after validation-based model selection. Evaluation verifies the saved split contents and writes accuracy, per-class precision/recall/F1/support, confusion matrices and training curves. Do not tune against the test results.


```python
if RUN_EVALUATION:
    if not DATA_DIR:
        raise ValueError("Set SCENE_DATA_DIR first")
    results = evaluate(DATA_DIR, RUN_DIR, BATCH_SIZE)
    for model, result in results.items():
        print(model, "test accuracy:", result["accuracy"])
else:
    print("Evaluation disabled; no new test metrics claimed.")
```
